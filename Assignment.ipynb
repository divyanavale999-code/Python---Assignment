```python
def returnValue(n1, n2):

    if(n1 == n2):
        return True
    elif(n1 - n2 == 5):
        return True
    elif(n1 + n2 == 5):
        return True

n1 = int(input("Enter first number : "))
n2 = int(input("Enter second number : "))

result = returnValue(n1, n2)
print(result)
```

```python
text = "The quick brown fox jumps over the lazy dog"

def countCharacters(input_string):
    frequency = {}

    for char in input_string:
        if char in frequency:
            frequency[char] = frequency[char] + 1
        else:
            frequency[char] = 1
    return frequency

result = countCharacters(text)
print(result)
```

```python
sentence = 'apple banana apple cherry banana apple'

words = sentence.split()

word_count = {}

for word in words:
    word_count[word] = word_count.get(word, 0) + 1

print(word_count)
```

```python
def swappingString(str1, str2):
    new_string = str2[ : 2] + str1[2 : ]
    new_str = str1[ : 2] + str2[2 : ]
    return new_string + " " + new_str

swappingString("Divya", "Vidhi")
```

```python
def changeString(string):
    if len(string) < 3:
        return string
    if string.endswith('ing'):
        return string + 'ly'
    else:
        return string + 'ing'

result = changeString("shivani")
print(result)
```
